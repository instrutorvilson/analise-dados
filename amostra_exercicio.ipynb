```python
#5 gerar 60% mulheres 40% homens
import pandas as pd
import numpy as np

np.random.seed(42)

# Tamanho da população
n_total = 500

# Definir proporções
n_mulheres = int(n_total * 0.6)  # 60% mulheres
n_homens = n_total - n_mulheres  # 40% homens

# Criar população
populacao = pd.DataFrame({
    'id': range(1, n_total + 1),
    'sexo': ['F']*n_mulheres + ['M']*n_homens
})

# Amostragem estratificada proporcional
amostra = pd.concat([
    populacao[populacao['sexo']=='F'].sample(n=30, random_state=42),
    populacao[populacao['sexo']=='M'].sample(n=20, random_state=42)
])
print(amostra['sexo'].value_counts())

```

```python
import pandas as pd

# Criar população
populacao = pd.DataFrame({
    'id': range(1, 401),
    'Setor': ['Financeiro']*60 + ['RH']*40 + ['TI']*100 + ['Vendas']*200
})

# Tamanho da amostra desejada
n_amostra = 40

# Calcular proporção de cada setor
proporcao = populacao['Setor'].value_counts(normalize=True)
# Número de pessoas de cada setor na amostra
n_por_setor = (proporcao * n_amostra).round().astype(int)

print("Número de pessoas por setor na amostra:")
print(n_por_setor)
```

```python
#9
import pandas as pd
import numpy as np

# Criar o DataFrame da população
populacao = pd.DataFrame({
    'ID': [1, 2, 3, 4, 5, 6],
    'Idade': [18, 21, 25, 29, 31, 45]
})

# Definir tamanho da amostra
n_amostra = 3

# -------------------------------
# a) Amostragem Aleatória Simples (AAS)
# -------------------------------
aas = populacao.sample(n=n_amostra, random_state=42)
print("Amostragem Aleatória Simples:")
print(aas)

# -------------------------------
# b) Amostragem Sistemática
# -------------------------------
N = len(populacao)
k = N // n_amostra  # intervalo
inicio = 1  # podemos começar do primeiro elemento
indices_sistematica = list(range(inicio, N, k))[:n_amostra]  # pegar n elementos
sistematica = populacao.iloc[indices_sistematica]
print("\nAmostragem Sistemática:")
print(sistematica)

# -------------------------------
# c) Amostragem Estratificada
# Estratos: Jovens (<30), Adultos (>=30)
# -------------------------------
# Criar coluna de estrato
populacao['Estrato'] = np.where(populacao['Idade'] < 30, 'Jovens', 'Adultos')

# Número de pessoas por estrato proporcional
n_jovens = round(n_amostra * (populacao['Estrato'].value_counts()['Jovens']/N))
n_adultos = n_amostra - n_jovens

# Amostragem estratificada
estratificada = pd.concat([
    populacao[populacao['Estrato']=='Jovens'].sample(n=n_jovens, random_state=42),
    populacao[populacao['Estrato']=='Adultos'].sample(n=n_adultos, random_state=42)
])

print("\nAmostragem Estratificada:")
print(estratificada)
```

```python
# 13 População com Estratos Desbalanceados
import pandas as pd

# Criar população
populacao = pd.DataFrame({
    'id': range(1, 371),
    'serie': ['1º ano']*300 + ['2º ano']*50 + ['3º ano']*20
})

# Tamanho da amostra
n_amostra = 37

# Amostragem proporcional
# Calcular proporção de cada série
proporcao = populacao['serie'].value_counts(normalize=True)
n_por_serie_proporcional = (proporcao * n_amostra).round().astype(int)

# Amostra proporcional
amostra_proporcional = populacao.groupby('serie', group_keys=False).apply(
    lambda x: x.sample(n=n_por_serie_proporcional[x.name], random_state=42)
)

print("Amostragem Proporcional:")
print(amostra_proporcional['serie'].value_counts())

# Amostragem não proporcional
# Escolher, por exemplo, 10 alunos de cada série
#n_por_serie_nao_proporcional = {'1º ano':10, '2º ano':10, '3º ano':10}

amostra_nao_proporcional = pd.concat([
    populacao[populacao['serie']=='1º ano'].sample(n=10, random_state=42),
    populacao[populacao['serie']=='2º ano'].sample(n=10, random_state=42),
    populacao[populacao['serie']=='3º ano'].sample(n=10, random_state=42),
])

print("\nAmostragem Não Proporcional:")
print(amostra_nao_proporcional['serie'].value_counts())
```
