```python
df["data_agendamento_limpa"] = pd.to_datetime(
    df["data_agendamento"],
    errors='coerce'
)
```

```python
df
```

```python
df['data_agendamento'] = df['data_agendamento'].str.replace('/', '-')
```

```python
# ajustar turnos
map_turno = {
    "Manha": "Manhã",
    "Manhã": "Manhã",
    "manha": "Manhã",
    "Tarde": "Tarde",
    "Noite": "Noite",
    "None": None
}
df["turno"] = df["turno"].map(map_turno)
```

```python
#preencher com valor padrão
df["especialidade"].fillna("Não Informado", inplace=True)
df["medico"].fillna("Não Informado", inplace=True)
#df["turno"].fillna("Não Informado", inplace=True)
```

```python
df
```

```python
#criar campos uteis para analise
#df["data_agendamento_limpa"].dropna(inplace=True)
df.dropna(subset=['data_agendamento_limpa'], inplace = True)
df["ano"] = df["data_agendamento_limpa"].dt.year
df["mes"] = df["data_agendamento_limpa"].dt.month
df["dia"] = df["data_agendamento_limpa"].dt.day
df["dia_semana"] = df["data_agendamento_limpa"].dt.day_name()
```
