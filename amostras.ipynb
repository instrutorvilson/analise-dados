```python
# amostragem estratificada
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'id': range(1, 501),
    'sexo': ['M']*300 + ['F']*200
})

amostra = df.groupby('sexo').sample(frac=0.2, random_state=42)  # 10% de cada grupo

```
